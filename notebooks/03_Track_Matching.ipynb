# Goal

Determine whether replacement tracks can be reliably paired with their original tracks using exported metadata alone.

## Initial assumptions

- Track ID cannot be used.
- Persistent ID cannot be used.
- Total Time is descriptive rather than identifying.
- Name, Artist and Album are expected to identify the musical work.
- A replacement track is expected to have Track Type = Remote or to have originated as Remote.

```python
results = {}
test_tracks = [
    ("Strength", "Strength 1985-1986", "The Alarm"),
    ("Blaze Of Glory", "Declaration", "The Alarm"),
    ("Hysteria", "H.A.A.R.P [Live]", "Muse"),
    ("Shoot To Thrill", "Back In Black", "AC/DC")
]

for track in test_tracks:
    name, album, artist = track
    track_series = baseline_df[(baseline_df['Name'] == name) & (baseline_df['Album'] == album) & (baseline_df['Artist'] == artist)].iloc[-1]
    results[name] = find_candidate_originals(track_series, original_pool)
    print(results[name])
```

```python
test_tracks = current_df[(current_df['Track Type'] == 'File') & (current_df['Comments'] == 'Owner: Steve')]
```

```python
test_tracks.shape
```

```python
results = []
for _, track in test_tracks.iterrows():
    candidates = find_candidate_originals(track, original_pool)
    results.append({"Track ID": track['Track ID'],
                    "Name": track['Name'],
                    "Artist": track['Artist'],
                    "Album": track['Album'],
                    "Candidates": len(candidates)
                   })
results_df = pd.DataFrame(results)
```

```python
one_candidate = results_df[results_df['Candidates'] == 1][['Name', 'Artist', 'Album']].sort_values(by=['Artist', 'Album', 'Name'])
```

```python
identity_fields = ['Name', 'Artist', 'Album', 'Track Number']
identity_df = current_df.sort_values(by='Track ID').copy()
identity_df['Identity Count'] = (
    identity_df
    .groupby(identity_fields)['Track ID']
    .transform('count')
)
identity_df['Max Track ID'] = (
    identity_df
        .groupby(identity_fields)['Track ID']
        .transform('max')
)
identity_df.iloc[50:60]
```
