```python
import xml.etree.ElementTree as ET
from pathlib import Path
```

```python
xml_path = "/Users/stephenusher/Documents/Steve's Folder/music-library-tools/exports/raw/"
xml_file = xml_path + "MusicLibrary_2026_08_03.xml"
```

```python
tree = ET.parse(xml_file)
root = tree.getroot()
```

```python
for i, child in enumerate(root):
    print(f"Child {i}")
    print(f"Tag: {child.tag}")
    print(f"Children: {len(child)}")
```

```python
library = root[0]

for i in range(0, len(library), 2):
    key = library[i]
    value = library[i + 1]
    
    print(f"{i // 2 + 1:2}. {key.text:20} -> {value.tag}")
```

```python
tracks = None

for i in range(0, len(library), 2):
    if library[i].text == "Tracks":
        tracks = library[i + 1]
        break
```
