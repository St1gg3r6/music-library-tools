```python
row_count = len(df)
profile = []
for column in df.columns:
    present = df[column].count()
    missing = row_count - present
    first_value = df[column].dropna().iloc[0]
    col_type = type(first_value).__name__
    profile.append({
        'Column': column,
        'Type': col_type,
        'Present': present,
        'Missing': missing,
        'Percentage Present': round(100 * present / row_count, 2)
    })
field_profile = pd.DataFrame(profile).sort_values(by='Present', ascending=False)
print(field_profile.to_markdown(index=False))
```

# Case Study: Known Removed Tracks

```python
def compare_replacements(name, album, artist):
    fields = ['Artist', 'Album', 'Name', 'Track ID', 'Persistent ID', 'Kind', 'Track Type', 'Location', 'Play Count', 'Rating']
    comparison = df[(df['Name'] == name) & (df['Album'] == album) & (df['Artist'] == artist)][fields].T
    comparison.columns = ['Original', 'Replacement']
    return comparison.to_markdown()
```

```python
name = 'Hysteria'
album = 'H.A.A.R.P [Live]'
artist = 'Muse'
print(compare_replacements(name, album, artist))
```

```python
name = 'Blaze Of Glory'
album = 'Declaration'
artist = 'The Alarm'
print(compare_replacements(name, album, artist))
```

```python
name = 'Marching On'
album = 'Declaration'
artist = 'The Alarm'
print(compare_replacements(name, album, artist))
```

```python
def compare_differences(name, album, artist):
    comparison = df[(df['Name'] == name) & (df['Album'] == album) & (df['Artist'] == artist)].T
    assert len(comparison.columns) == 2, ('Expected exactly 2 matching tracks.')
    comparison.columns = ['Original', 'Replacement']
    mask = ~(
        comparison['Original'].isna() & comparison['Replacement'].isna()
    ) & (
        comparison['Original'] != comparison['Replacement']
    )
    comparison = comparison[mask]
    print(f"There were {len(comparison)} fields changed in the replacement track.")
    return comparison
```

```python
name = 'Hysteria'
album = 'H.A.A.R.P [Live]'
artist = 'Muse'
print(compare_differences(name, album, artist).to_markdown())
```

```python
name = 'Blaze Of Glory'
album = 'Declaration'
artist = 'The Alarm'
print(compare_differences(name, album, artist).to_markdown())
```

```python
name = 'Marching On'
album = 'Declaration'
artist = 'The Alarm'
print(compare_differences(name, album, artist).to_markdown())
```
